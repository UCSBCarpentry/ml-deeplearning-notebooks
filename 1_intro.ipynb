## Two Meta-Problems in Machine Learning
### 1. Estimation
We want to make estimates on a quantity of interest, by aggregating some noisy reflections
### 2. Generalization
We want to make sure our prediction works on a situation never seen in the data set

## How do we go about describeing a ML task? Use these six characteristics:
- Problem class: What is the nature of the training data and what kinds of queries will be made at testing time?
- Assumptions: What do we know about the source of the data or the form of the solution?
- Evaluation criteria: What is the goal of the prediction or estimation system? How will the answers to individual queries be evaluated? How will the overall performance of the system be measured?
- Model type: What aspects of the data will be modeled? How do we use this model to make predictions?
- Model class: What parametric class of models will be used? How do we pick a particular model from this class?
- Algorithms: How to fit the model to data?
  

## Common Problem Classes

### 1. Supervised Learning
The learning system is given: **inputs** and **expected outputs** 
It can be further divided to:
- Classification
- Regression

### Classification Problem
- Data: Pairs of (input, output)
- Input: Input represents a description of the object to the classified.
- Output: An element from a finite, unordered set.

### Regression Problem
- Data: Same as Classification
- Input: Same as Classification
- Output: A continuous value/vector.

### Examples of Classification Problem
- Input: The image of an animal
- Output: The species of this animal. Target space is unordered set of 5000 species

- Input: The facial image of a person, captured at a facial recognition gate.
- Output: The employer ID in an internal database.
### To summarize, a classification model places input data into a category / bucket. There is no notion of distance between buckets.


### Examples of Regression Problem
- Input: The image of an cat
- Output: The weight of the cat. 

- Input: The facial image of a person
- Output: The age of the person.
### To summarize, a regression model predicts a numerical value. The value lives in a space with distance measures.

### 2. Unsupervised Learning
Unsupervised learning does not learn a function that maps a given input to the given output.
Instead, one is given a data set and generally expected to **find some patterns or structure inherent in it**.

### Clustering
Given samples, find a partitioning/clustering such that samples in the same group are **similar**

### Clustering Example: Social Media Community Detection
- input: Social media activities of a set of users (likes, comments, following...)
- output: A subset of users that are close / know each other.
- how is this useful? It can be used to recommend people you may know to you.
- Why is this unsupervised learning: There is no label to flag **different friend groups** in social media data

### Dimensionality Reduction
Given samples described with D variables (or samples in D-dimensional space).
Find a mapping that represent the samples with only d variables (d<D), such that elements in **different groups** are still separated. 

### Dimensionality Reduction Example: Faster Database Navigation
- the problem: A patient database has millions of patients, and hundres of features per patient. We need to find patients with similar conditions.
- the machine learning approach: find a projection that maps the features of each patient to a low-dimensional space. Say only 10 features.
- how is this useful: We can reduce computation cost by search for similar patients in the low-dimensional space.
- Why is this unsupervised learning: There is no ground truth to tell us how to project.

## 3. Prepare data

(Most) Machine learning libraries only take numerical values without missing (NaN) values, what can we do?
- Drop categorical/text features and all rows with missing values (at the price of losing valuable data)
- Transform categorical features to numerical. Most simple method: one-hot encoding (also called dummy variable encoding)
- Fill missing values (imputation). Simple: Replace NaN with the average of the feature

### Prepare features

```python
# Use only numerical features: bill_length_mm, bill_depth_mm, flipper_length_mm, body_mass_g
## That's the same as dropping the categorical features
penguins_filtered = penguins.drop(columns=['island', 'sex'])

# Drop rows with NaNs
penguins_filtered = penguins_filtered.dropna()

# Our features are ready, we will keep all but our target variable 'species'
features = penguins_filtered.drop(columns=['species'])
```

```python
# Extract columns corresponding to features - this is our prediction target!
features = penguins_filtered.drop(columns=['species'])
```

### Prepare target variable

```python
# By default, the target variable is a text string:
penguins_filtered['species']
```

```python
# We need to convert to numerical 
target, cat_indices = pd.factorize(penguins_filtered['species'])
print(target)
print(cat_indices)
```

```python
# dataset split
X_train, X_test, y_train, y_test = train_test_split(features, target, test_size=0.2, random_state=0, shuffle=True, stratify=target)
```

```python
classifier = LogisticRegression(random_state=0, max_iter=200, C=np.inf)
classifier.fit(X_train, y_train)
print(classifier.score(X_train, y_train))
print(classifier.score(X_test, y_test))
```

```python
# Understand the classifier's parameters
print(classifier.coef_.shape)
print(classifier.intercept_.shape)
```
